```python
from PIL import Image
import os
from tqdm import tqdm
import numpy as np
import shutil
from sklearn.model_selection import train_test_split

from PIL import ImageFile
ImageFile.LOAD_TRUNCATED_IMAGES = True
```

```python
real_source = '/media/data2/dataset/GAN_ImageData/star/0'
fake_source = '/media/data2/dataset/GAN_ImageData/star/1'
real_category = ['train', 'validation', 'test']
target = '/media/data2/dataset/GAN_ImageData/StarGAN_128/'
```

```python
real_list = os.listdir(real_source)
```

```python
real_train, real_test = train_test_split(real_list, test_size=50000)
real_train, real_val = train_test_split(real_train, test_size=0.1)
```

```python
real_set = [real_train, real_val, real_test]
```

```python
fake_list = os.listdir(fake_source)
fake_list = np.asarray(fake_list)
```

```python
for i in range(3):
    r_set = real_set[i]
    for r in tqdm(r_set):
        real_source_dir = os.path.join(real_source, r)
        target_folder_dir = os.path.join(target, real_category[i])
        real_target_dir = os.path.join(target_folder_dir, '0')
        fake_target_dir = os.path.join(target_folder_dir, '1')
        if not os.path.isdir(real_target_dir):
            os.makedirs(real_target_dir)
        if not os.path.isdir(fake_target_dir):
            os.makedirs(fake_target_dir)
        real_target_img = os.path.join(real_target_dir, r)
        
        rn = np.random.randint(low=1, high=6, size=1)
        target_rn = r[:-4]+'_'+str(rn[0])+'.jpg'
        fake_img = fake_list[fake_list==target_rn][0]
        fake_source_dir = os.path.join(fake_source, fake_img)
        fake_img_dir = os.path.join(fake_target_dir, fake_img)
        
        shutil.copy2(real_source_dir, real_target_img)
        shutil.copy2(fake_source_dir, fake_img_dir)
```
